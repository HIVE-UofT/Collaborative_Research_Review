```python
import pandas as pd
from itertools import combinations

# Load the data
file_path = 'LLama3.1_Feature_Extraction_Accuracy_Check - patient-level-ds.csv'  # Replace with your actual file path
canada_df = pd.read_csv(file_path)

# Ensure that the 'countries' column contains lists (if stored as strings in the CSV)
canada_df['countries'] = canada_df['countries'].apply(eval)

# Extract collaborations involving Canada
collaborations = canada_df['countries'].apply(lambda x: [(a, b) for a, b in combinations(x, 2) if 'Canada' in [a, b]])

# Flatten the list of lists
collaborations = [pair for sublist in collaborations for pair in sublist]

# Filter to keep only the collaborations where Canada is involved
canada_collaborations = [pair for pair in collaborations if 'Canada' in pair]

# Convert pairs to a DataFrame
canada_collaborations_df = pd.DataFrame(canada_collaborations, columns=['Country 1', 'Country 2'])

# Ensure that 'Canada' is always in the first column and the other country in the second
canada_collaborations_df['Other Country'] = canada_collaborations_df.apply(lambda row: row['Country 2'] if row['Country 1'] == 'Canada' else row['Country 1'], axis=1)

# Count the occurrences of each country
country_counts = canada_collaborations_df['Other Country'].value_counts()

# List the top 15 countries collaborating with Canada
top_15_countries = country_counts.head(15)

# Calculate the total number of collaborations
total_collaborations = canada_df.shape[0]

# Calculate the percentage for each of the top 15 countries
top_15_countries_df = top_15_countries.reset_index()
top_15_countries_df.columns = ['Country', 'Collaboration Count']
top_15_countries_df['Percentage'] = (top_15_countries_df['Collaboration Count'] / total_collaborations) * 100

# Display the final DataFrame
print(top_15_countries_df)
```

```python
import pandas as pd
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
import re

# Initialize the stemmer
stemmer = PorterStemmer()

# Filter the countries with more than 5% collaboration
top_countries_above_5_percent = top_15_countries_df[top_15_countries_df['Percentage'] > 5]['Country']

# Filter the original DataFrame to get rows where the collaborating countries have more than 5% collaboration
filtered_canada_df = canada_df[canada_df['countries'].apply(lambda x: any(country in top_countries_above_5_percent.values for country in x))]


# Function to normalize and stem text
def normalize_text(text):
    text = text.lower()  # Convert to lowercase
    text = re.sub(r'[^\w\s]', '', text)  # Remove non-alphanumeric characters except for whitespace
    tokens = word_tokenize(text)  # Tokenize
    normalized_tokens = [stemmer.stem(token) for token in tokens]  # Stem
    return ' '.join(normalized_tokens)

# Function to process and combine Modalities and Modalities2 columns
def process_modalities(row):
    combined = (row['Modalities'] + ',' + row['Modalities2']).replace('""', '"')
    modalities_list = [normalize_text(name.strip()) for name in combined.split(',')]
    return list(set(modalities_list))  # Remove duplicates and return as a list

# Apply the function to the filtered DataFrame
filtered_canada_df['unique_modalities'] = filtered_canada_df.apply(process_modalities, axis=1)
```

```python
import os
import json

def extract_json_objects_from_text(text):
    json_objects = []
    while True:
        # Find the first '{' and the last '}'
        start_index = text.find('{')
        end_index = text.rfind('}')

        if start_index == -1 or end_index == -1 or start_index > end_index:
            break

        # Extract the substring from the first '{' to the last '}'
        json_str = text[start_index:end_index + 1]

        # Try to parse the extracted substring as JSON
        try:
            json_object = json.loads(json_str)
            json_objects.append(json_object)
            # Remove the processed part of the text
            text = text[end_index + 1:]
        except json.JSONDecodeError:
            # If parsing fails, continue searching in the remaining text
            text = text[end_index + 1:]

    return json_objects

def extract_json_objects_from_txt(file_path):
    with open(file_path, 'r') as file:
        content = file.read()
    return extract_json_objects_from_text(content)

def read_json_from_folder(folder_path):
    all_json_objects = []
    for file_name in os.listdir(folder_path):
        if file_name.endswith('.txt'):
            file_path = os.path.join(folder_path, file_name)
            json_objects = extract_json_objects_from_txt(file_path)
            # If no JSON objects were found, add an empty object
            if not json_objects:
                json_objects.append({})
            # Add the JSON objects with the corresponding PID (file name without extension)
            for json_object in json_objects:
                pid = os.path.splitext(file_name)[0]
                all_json_objects.append({'pid': pid, 'data': json_object})
    return all_json_objects

# Example usage
folder_path = './perplexity-ft2/'
json_array = read_json_from_folder(folder_path)
print(json_array)
```

```python
import os
import json

def extract_json_objects_from_text(text):
    json_objects = []
    while True:
        # Find the first '{' and the last '}'
        start_index = text.find('{')
        end_index = text.rfind('}')

        if start_index == -1 or end_index == -1 or start_index > end_index:
            break

        # Extract the substring from the first '{' to the last '}'
        json_str = text[start_index:end_index + 1]

        # Try to parse the extracted substring as JSON
        try:
            json_object = json.loads(json_str)
            json_objects.append(json_object)
            # Remove the processed part of the text
            text = text[end_index + 1:]
        except json.JSONDecodeError:
            # If parsing fails, continue searching in the remaining text
            text = text[end_index + 1:]

    return json_objects

def extract_json_objects_from_txt(file_path):
    with open(file_path, 'r') as file:
        content = file.read()
    return extract_json_objects_from_text(content)

def read_json_from_folder(folder_path):
    all_json_objects = []
    for file_name in os.listdir(folder_path):
        if file_name.endswith('.txt'):
            file_path = os.path.join(folder_path, file_name)
            json_objects = extract_json_objects_from_txt(file_path)
            # If no JSON objects were found, add an empty object
            if not json_objects:
                json_objects.append({})
            # Add the JSON objects with the corresponding PID (file name without extension)
            for json_object in json_objects:
                pid = os.path.splitext(file_name)[0]
                all_json_objects.append({'pid': pid, 'data': json_object})
    return all_json_objects

# Example usage
folder_path = './chatgpt/ok'
json_array = read_json_from_folder(folder_path)
print(json_array)
```
